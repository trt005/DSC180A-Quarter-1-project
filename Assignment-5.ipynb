```python
import os
import urllib.request
import numpy as np
import pandas as pd
```

```python

url = 'https://f000.backblazeb2.com/file/jeldridge-data/mnist.npz'
output_dir = './localstorage'
local_filepath = os.path.join(output_dir, 'mnist.npz')

if not os.path.exists(local_filepath):
    urllib.request.urlretrieve(url, local_filepath)

with np.load(local_filepath) as data:
    X_train = data['train'].T
    y_train = data['train_labels'].T
    X_test = data['test'].T
    y_test = data['test_labels'].T

# Generate header column name
pixel_columns = [f"pixel_{i}" for i in range(784)]

# Build structured Training DataFrame
df_train = pd.DataFrame(X_train, columns=pixel_columns)
df_train['label'] = y_train.astype(int)

# Build structured Testing DataFrame
df_test = pd.DataFrame(X_test, columns=pixel_columns)
df_test['label'] = y_test.astype(int)

print(f"\nSuccessfully loaded into DataFrames!")
print(f"Training shape: {df_train.shape} | Testing shape: {df_test.shape}")
```
